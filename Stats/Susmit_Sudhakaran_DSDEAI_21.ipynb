```python
import pandas as pd
import numpy as np
from scipy import stats
from math import comb
```

```python
# Load data
df = pd.read_csv("property.csv")

# Preview
print(df.head())
print(df.columns)
```

For the suburb of Altona, it is postulated that a typical property sells for 800,000. Use the data at hand to test this assumption. Is the typical property price really $800,000 or has it increased? Use a significance level of 5%.


```python
altona = df[df['Suburb'].str.lower() == 'altona']
price = altona['Price'].dropna()

# H0: mean = 800000
# H1: mean > 800000
t_stat, p_val = stats.ttest_1samp(price, 800000)

print("Q1: Altona Property Price Test")
print(f"t-statistic = {t_stat:.4f}, p-value = {p_val/2:.4f}")

if (p_val/2 < 0.05) and (t_stat > 0):
    print("Reject H0 → Average price has significantly increased above $800,000.")
else:
    print("Fail to Reject H0 → No significant increase from $800,000.")
```

For the year 2016, is there any difference in the prices of properties sold in the
summer months vs winter months? Consider months from October till March
as winter months and rest as summer months. Use a significance level of 5%.


```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df_2016 = df[df['Date'].dt.year == 2016]

# Define seasons 
df_2016['Month'] = df_2016['Date'].dt.month
winter = df_2016[df_2016['Month'].isin([10, 11, 12, 1, 2, 3])]
summer = df_2016[~df_2016['Month'].isin([10, 11, 12, 1, 2, 3])]

t_stat2, p_val2 = stats.ttest_ind(winter['Price'].dropna(), summer['Price'].dropna(), equal_var=False)

print("Q2: 2016 – Winter vs Summer Property Prices")
print(f"t-statistic = {t_stat2:.4f}, p-value = {p_val2:.4f}")

if p_val2 < 0.05:
    print("Significant difference between winter and summer property prices.")
else:
    print("No significant difference between winter and summer property prices.")
```

For the suburb of Abbotsford, what is the probability that out of 10 properties
sold, 3 will not have a car parking space? Use the column car in the dataset.
Round off your answer to 3 decimal places

```python
abbotsford = df[df['Suburb'].str.lower() == 'abbotsford']
no_car_prob = np.mean(abbotsford['Car'].fillna(0) == 0)

# Binomial: P(X=3) when n=10, p=no_car_prob
n, k = 10, 3
binom_prob = comb(n, k) * (no_car_prob**k) * ((1 - no_car_prob)**(n - k))

print(f"Q3: P(3 of 10 Abbotsford properties have no car space) = {binom_prob:.3f}")
```

In the suburb of Abbotsford, what are the chances of finding a property with 3
rooms? Round your answer to 3 decimal places.

```python
total_properties = len(abbotsford)

three_room_properties = len(abbotsford[abbotsford['Rooms'] == 3])

p_3rooms = three_room_properties / total_properties
print(f"Probability of 3-room property in Abbotsford = {p_3rooms:.3f}")
```

In the suburb of Abbotsford, what are the chances of finding a property with 2
bathrooms? Round your answer to 3 decimal places.


```python
two_bath_properties = len(abbotsford[abbotsford['Bathroom'] == 2])

p_2bath = two_bath_properties / total_properties
print(f"Probability of 2-bathroom property in Abbotsford = {p_2bath:.3f}")
```
