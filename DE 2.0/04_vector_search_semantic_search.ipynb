## 🔎 Vector Search & Semantic Retrieval – SmartGear

### 🎯 Objective

To implement a semantic search system using product metadata by converting text into vector embeddings and retrieving similar results based on meaning rather than exact keyword matching.

---

### 🧠 Approach

1. Created a product metadata dataset
2. Generated embeddings using Sentence Transformers
3. Stored embeddings in a vector database (ChromaDB)
4. Implemented semantic search using similarity matching

---

### 📊 Dataset Schema

* product_id
* product_name
* category
* description

---

### 🚀 Key Concepts

* **Embeddings**: Numerical representation of text
* **Vector Database**: Stores embeddings for similarity search
* **Semantic Search**: Finds results based on meaning

---

### 📌 Outcome

This implementation enables:

* Intelligent product search
* Better user experience
* AI-powered recommendations

```python
from sentence_transformers import SentenceTransformer
import chromadb
```

```python
#Creating Product Meta Data
products = [
    {"product_id": 1, "product_name": "Headphones", "category": "Accessories", "description": "Wireless noise cancelling headphones for music and calls"},
    {"product_id": 2, "product_name": "Smartwatch", "category": "Wearables", "description": "Smartwatch with fitness tracking and notifications"},
    {"product_id": 3, "product_name": "Printer", "category": "Electronics", "description": "High speed printer for office and home use"},
    {"product_id": 4, "product_name": "Camera", "category": "Electronics", "description": "Digital camera with high resolution and zoom"},
    {"product_id": 5, "product_name": "Tablet", "category": "Electronics", "description": "Portable tablet for entertainment and productivity"},
    {"product_id": 6, "product_name": "Smartphone", "category": "Electronics", "description": "Latest smartphone with advanced features and camera"},
    {"product_id": 7, "product_name": "Drone", "category": "Electronics", "description": "Flying drone with camera for aerial photography"},
    {"product_id": 8, "product_name": "Monitor", "category": "Electronics", "description": "HD monitor for gaming and professional work"},
    {"product_id": 9, "product_name": "Gaming Console", "category": "Gaming", "description": "Console for immersive gaming experience"},
    {"product_id": 10, "product_name": "Laptop", "category": "Electronics", "description": "High performance laptop for work and gaming"}
]
```

```python
#Generating Embeddings
model = SentenceTransformer('all-MiniLM-L6-v2')

for product in products:
    product["embedding"] = model.encode(product["description"]).tolist()
```

```python
#Storing in Vector DB
client = chromadb.Client()
collection = client.create_collection(name="smartgear_products")

for product in products:
    collection.add(
        ids=[str(product["product_id"])],
        documents=[product["description"]],
        embeddings=[product["embedding"]],
        metadatas=[{
            "product_name": product["product_name"],
            "category": product["category"]
        }]
    )
```

```python
# Semantic Search Function
def search_products(query):
    query_embedding = model.encode(query).tolist()
    
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=3
    )
    
    return results
```

```python
query = "best device for gaming"
results = search_products(query)
```

```python
for i in range(len(results['documents'][0])):
    print("Result:", i+1)
    print("Description:", results['documents'][0][i])
    print("Metadata:", results['metadatas'][0][i])
    print("-" * 50)
```
